```python
"""
Smart Electric Vehicle Battery Performance Analysis
Case Study Solution — Battery Data Analytics Team
Tools: Python, NumPy, Linear Algebra, Calculus
"""

import numpy as np

print("=" * 65)
print("TASK 1 — ANALYSIS USING PYTHON BASICS")
print("=" * 65)

# ---- Store sensor readings ----
time_h       = [0, 1, 2, 3, 4, 5]
temperature  = [32, 35, 38, 40, 42, 45]
voltage      = [400, 395, 390, 385, 380, 375]
current      = [20, 22, 25, 28, 30, 32]
charge       = [95, 88, 80, 70, 60, 48]

# 1. Display the sensor readings
print("\n1. Sensor Readings:")
print(f"{'Time(h)':>8} {'Temp(C)':>8} {'Volt(V)':>8} {'Curr(A)':>8} {'Charge(%)':>10}")
for i in range(len(time_h)):
    print(f"{time_h[i]:>8} {temperature[i]:>8} {voltage[i]:>8} {current[i]:>8} {charge[i]:>10}")

# 2. Identify high temperature
print("\n2. High Temperature Check (T > 40 C):")
for i in range(len(time_h)):
    if temperature[i] > 40:
        print(f"  t={time_h[i]}h, T={temperature[i]}C -> High Temperature Alert")
    else:
        print(f"  t={time_h[i]}h, T={temperature[i]}C -> Temperature Normal")

# 3. Identify low battery
print("\n3. Low Battery Check (C < 50%):")
for i in range(len(time_h)):
    if charge[i] < 50:
        print(f"  t={time_h[i]}h, C={charge[i]}% -> Low Battery Alert")
    else:
        print(f"  t={time_h[i]}h, C={charge[i]}% -> Battery Charge Normal")

# 4. battery_status function
def battery_status(charge):
    if charge > 70:
        return "High"
    elif charge >= 50:
        return "Medium"
    else:
        return "Low"

print("\n4. Battery Status Function:")
for i in range(len(time_h)):
    print(f"  t={time_h[i]}h, C={charge[i]}% -> Status: {battery_status(charge[i])}")


print("\n" + "=" * 65)
print("TASK 2 — ANALYSIS USING NUMPY")
print("=" * 65)

D = np.array([
    [32, 400, 20, 95],
    [35, 395, 22, 88],
    [38, 390, 25, 80],
    [40, 385, 28, 70],
    [42, 380, 30, 60],
    [45, 375, 32, 48]
])

# 1. Shape of the array
print(f"\n1. Shape of array D: {D.shape}  -> {D.shape[0]} rows, {D.shape[1]} columns")

# 2. Average of each parameter
T_bar = D[:, 0].mean()
V_bar = D[:, 1].mean()
I_bar = D[:, 2].mean()
C_bar = D[:, 3].mean()
print("\n2. Averages:")
print(f"  Average Temperature (T_bar) = {T_bar:.2f} C")
print(f"  Average Voltage     (V_bar) = {V_bar:.2f} V")
print(f"  Average Current     (I_bar) = {I_bar:.2f} A")
print(f"  Average Charge      (C_bar) = {C_bar:.2f} %")

# 3. Max and min temperature
T_max = D[:, 0].max()
T_min = D[:, 0].min()
print(f"\n3. T_max = {T_max} C, T_min = {T_min} C")

# 4. Extract voltage column
V_col = D[:, 1]
print(f"\n4. Voltage column V = {V_col}")

# 5. High-temperature rows (T > 40)
high_temp_rows = D[D[:, 0] > 40]
print(f"\n5. Rows with T > 40 C:\n{high_temp_rows}")

# 6. Low battery rows (C < 50)
low_charge_rows = D[D[:, 3] < 50]
print(f"\n6. Rows with C < 50%:\n{low_charge_rows}")


print("\n" + "=" * 65)
print("TASK 3 — ANALYSIS USING LINEAR ALGEBRA")
print("=" * 65)

A = np.array([[2, 1],
              [1, 3]])
X = np.array([[40],
              [25]])

# Y = A X
Y = A @ X
print("\nBattery Stress Index Transformation Y = A X:")
print(f"A =\n{A}")
print(f"X =\n{X}")
print(f"Y =\n{Y}")

# 1. Transpose
A_T = A.T
print(f"\n1. Transpose A^T =\n{A_T}")

# 2. Determinant
det_A = np.linalg.det(A)
print(f"\n2. Determinant det(A) = {det_A:.2f}")

# 3. Inverse
if det_A != 0:
    A_inv = np.linalg.inv(A)
    print(f"\n3. Inverse A^-1 =\n{A_inv}")
else:
    print("\n3. Inverse does not exist (determinant = 0)")

# Solving system of equations: 2x + y = 100 ; x + 3y = 120
A_sys = np.array([[2, 1],
                   [1, 3]])
B_sys = np.array([100, 120])
solution = np.linalg.solve(A_sys, B_sys)
print(f"\nSolving A X = B for 2x + y = 100 and x + 3y = 120:")
print(f"  x = {solution[0]:.2f}")
print(f"  y = {solution[1]:.2f}")


print("\n" + "=" * 65)
print("TASK 4 — ANALYSIS USING CALCULUS")
print("=" * 65)

# C(t) = 100 - 2t - 0.5t^2
# dC/dt = -2 - t   (derivative)

def C(t):
    return 100 - 2*t - 0.5*t**2

def dC_dt(t):
    return -2 - t

print("\nC(t) = 100 - 2t - 0.5t^2")
print("dC/dt = -2 - t")

t1, t2 = 2, 5
rate_t1 = dC_dt(t1)
rate_t2 = dC_dt(t2)

print(f"\n1 & 2. Discharge rate:")
print(f"  dC/dt at t={t1} -> {rate_t1} %/hour")
print(f"  dC/dt at t={t2} -> {rate_t2} %/hour")

print("\n3. Interpretation:")
print("  The rate dC/dt = -2 - t is linear in t and becomes more negative")
print("  as t increases, meaning the battery is LOSING CHARGE AT AN")
print("  INCREASING RATE (accelerating discharge) over time.")


print("\n" + "=" * 65)
print("FINAL ANALYSIS AND DECISION — BATTERY PERFORMANCE REPORT")
print("=" * 65)

# 1. Safety - first time temp exceeds 40
first_exceed_idx = next(i for i, t in enumerate(temperature) if t > 40)
print(f"\n1. Safety: Temperature first exceeds 40 C at t = {time_h[first_exceed_idx]}h "
      f"(T = {temperature[first_exceed_idx]} C)")

# 2. Charge loss
charge_loss = charge[0] - charge[-1]
print(f"\n2. Battery Charge: Charge Loss = C_initial - C_final = "
      f"{charge[0]} - {charge[-1]} = {charge_loss}%")

# 3. NumPy Analysis
print(f"\n3. NumPy Analysis: T_bar={T_bar:.2f}C, V_bar={V_bar:.2f}V, "
      f"I_bar={I_bar:.2f}A, C_bar={C_bar:.2f}%")

# 4. Linear Algebra
print(f"\n4. Linear Algebra: Y = A X = [{Y[0,0]}, {Y[1,0]}]^T ; "
      f"Solution to A X = B -> x={solution[0]:.2f}, y={solution[1]:.2f}")

# 5. Calculus
print(f"\n5. Calculus: dC/dt at t=2 is {rate_t1} %/h, at t=5 is {rate_t2} %/h "
      f"-> discharge rate is increasing in magnitude over time")

# 6. Final decision
print("\n6. Final Decision: MONITOR")
print("   Temperature rises steadily above the 40C safety threshold from t=3h")
print("   onward, and charge depletes rapidly (47% lost over 5 hours) with an")
print("   accelerating discharge rate. The battery is not in immediate failure,")
print("   but trends are unfavorable and warrant continuous monitoring rather")
print("   than outright approval or rejection.")
```
