```python
import numpy as np
import sympy as sp

x, y = sp.symbols('x y')
sp.init_printing()
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)
```

```python
def f(x):
  return x**2

h=1e-6
slope_at_3=(f(3+h)-f(3))/h
print('Numerical f\'(3):',round(slope_at_3,4), '(exact=6)')
```

```python
def g(x):
  return x**3+2*x

h = 1e-5
x0 = 2
numerical_deriv = (g(x0 + h) - g(x0 - h)) / (2*h)
print("1. Numerical derivative at x=2:", numerical_deriv)

x = sp.symbols('x')
g_prime = sp.diff(g(x), x)
print("2. Symbolic derivative g'(x):", g_prime)

symbolic_at_2 = g_prime.subs(x, 2)
print("3. Symbolic derivative at x=2:", symbolic_at_2)
```

```python
f2=x**2 + 3*x*y + y**2
print('df/dx=',sp.diff(f2, x))
print('df/dy=',sp.diff(f2, y))
```

```python
grad=[sp.diff(f2,v) for v in [x,y]]
print('grad f=',grad)

grad_at=[g.subs({x:1,y:2}) for g in grad]
print('grad f at (1,2)=', grad_at)
```

```python
h2 = x**2 * y + sp.sin(y)

dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)

gradient = [dh_dx, dh_dy]

gradient_at_point = [comp.subs({x: 2, y: 0}) for comp in gradient]

print("dh/dx =", dh_dx)
print("dh/dy =", dh_dy)
print("Gradient:", gradient)
print("Gradient at (2, 0):", gradient_at_point)
```

```python
by_hand=sp.cos(x**2)*2*x
by_sympy=sp.diff(sp.sin(x**2),x)
print(by_hand)
print(by_sympy)
print('Match? ',sp.simplify(by_hand-by_sympy)==0)
```

```python
expr3=(3*x+1)**4
print('d/dx (3x+1)^4 =',sp.diff(expr3,x))
```

```python
# y = exp(x**2 + 1);  inner = x**2 + 1, inner' = 2x, outer' = exp(inner)

# 1. By hand (as a SymPy expression)
inner = x**2 + 1
inner_prime = 2*x
outer_prime = sp.exp(inner)
by_hand = outer_prime * inner_prime

# 2. With sp.diff
y = sp.exp(x**2 + 1)
sympy_diff = sp.diff(y, x)

# 3. Confirm they match
print("By hand:", by_hand)
print("sp.diff:", sympy_diff)
print("Match?", sp.simplify(by_hand - sympy_diff) == 0)
```

```python
f5 = x**2 + 3*x*y + y**2
H = sp.hessian(f5, (x, y))
print('Hessian of f:')
sp.pprint(H)
```
